```python
nome = input("digite o nome: ")
while len(nome) <= 3:
  nome = (input("nome inválido! digite um nome que tenha mais que 3 caracteres: "))

idade = int(input("digite sua idade: "))
while (idade) < 0 or idade > 150:
  idade = int(input("idade inválida! digite uma idade entre 0 a 150: "))

salario = float(input("digite seu salario: "))
while (salario) <= 0:
  salario = float(input("salario inválido! digite um salario 0 ou maior:"))

sexo = input("digite seu sexo (f/m): ").lower()
while sexo != "f" and sexo != "m":
  sexo = input("sexo inválido! digite um sexo entre f e m: ").lower()

estadocivil = input("digite seu estado civil s/c/v/d: ").lower()
while estadocivil not in ['s','c','v','d']:
 estadocivil = input ("estado civil inválido! digite um estado civil entre 's', 'c', 'v' e 'd': ").lower()

 print(f"nome: {nome}")
 print(f"idade: {idade}")
 print(f"salario: {salario}")
 print(f"sexo: {sexo}")
 print(f"estado civil: {estadocivil}")
print("Dados válidos!")
```

```python
n = int(input("digite um número inteiro: "))
primo = True

if n <= 1:
  primo = False
else:
  for i in range(2, n):
    if n % i == 0:
      primo = False
      break
if primo:
  print("é numero primo")
else:
  print("não é primo")
```

```python
numero = int(input("digite um numero inteiro: "))
FATORIAL = 1
for i in range(1, numero + 1):
  FATORIAL = FATORIAL * i

print(f"o fatorial do {numero} é: {FATORIAL}")
```

```python
L = [5,7,2,9,4,1,3]
tamanho = len(L)
maior = max(L)
menor = min(L)
soma = sum(L)
crescente = sorted(L)
decrescente = sorted(L, reverse=True)
print ("tamanho da lista:", tamanho)
print("maior valor:", maior)
print("menor valor:", menor)
print("a soma é:", soma)
print("ordem crescente:", crescente)
print("ordem decrescente:", decrescente)
```
